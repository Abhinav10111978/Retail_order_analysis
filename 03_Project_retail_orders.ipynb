### Level (1) : Basic analysis
     Total orders                       mark(✓)
     Total revenue                      mark(✓)
     Total quantity sold                mark(✓)
     Average order value                mark(✓)
     Delivered vs Cancelled orders      mark(✓)

```python
file = open("retail_orders_practice.csv", "r")
lines = file.readlines()

pure_data = []
for line in lines[1: ]:
    field = line.strip().split(",")
    field[6] = int(field[6])
    field[7] = float(field[7])
    field[8] = float(field[8])
    pure_data.append(field)
```

```python
# 1st Query : Total Orders

total_order = len(pure_data)
print(f"Total number of orders are : {total_order}")
```

```python
# 2nd Query : Total Revenue

total_revenue = 0

for row in pure_data:
    quantity = row[6]
    unit_price = row[7]
    discount = row[8]
    if discount == 0:
        total_revenue += quantity * unit_price 
    else:
        price_with_discount = unit_price - (unit_price * (discount/100))
        total_revenue += quantity * price_with_discount 

      
print(f"The total revenue is : ₹ {total_revenue:.3f}")
```

```python
# 3rd Query : Total Quantity

total_quantity = 0
category_quantity = {}

for row in pure_data:
    total_quantity += row[6]
    category = row[5]
    if category in category_quantity:
        category_quantity[category] += row[6]
    else:
        category_quantity[category] = row[6]


for ct, qun in category_quantity.items():
    print(f"{ct} : {qun}")
print(f"The entire quantities are : {total_quantity}")
```

```python
# 4th Query : Average order value

total_order = len(pure_data)
total_revenue = 0

for row in pure_data:
    quantity = row[6]
    unit_price = row[7]
    discount = row[8]
    if discount == 0:
        total_revenue += quantity * unit_price 
    else:
        price_with_discount = unit_price - (unit_price * (discount/100))
        total_revenue += quantity * price_with_discount 


avg_order_value = total_revenue/total_order
print(total_revenue)
print(f"The Average value of order is : ₹ {avg_order_value}")    
```

```python
# 5th Query : Total Order Delivered

Order_delivered = {}

for row in pure_data:
    status = row[10]
    if status in Order_delivered:
        Order_delivered[status] += +1
    else:
        Order_delivered[status] = +1

for st, cnt in Order_delivered.items():
    print(f"{st} : {cnt}")
```

### Level (2) : Dictionary mastery
    Revenue by city                         mark(✓)
    Revenue by category                     mark(✓)
    Total spending by each customer         mark(✓)
    Most popular payment method             mark(✓)
    Highest revenue-generating product      mark(✓)
    

```python
# 1st Query : Revenue by city

revenue_city = {}
revenue = 0
for row in pure_data:
    city = row[3]
    revenue = row[6]*row[7]
    
    if city in revenue_city:
        revenue_city[city] += revenue
    else:
        revenue_city[city] = revenue

for ct, rev in revenue_city.items():
    print(f"{ct} : ₹ {rev}")
```

```python
# 2nd Query : Revenue by category


category_revenue = {}

for row in pure_data:
    revenue = row[6]*row[7]
    category = row[5]

    if category in category_revenue:
        category_revenue[category] += revenue 
    else:
        category_revenue[category] = revenue

for ct, rev in category_revenue.items():
    print(f"{ct} : ₹ {rev}")
```

```python
# 3rd Query : Total spending by each customer

customer_spending = {}
money_revenue = 0

for row in pure_data:
    customer = row[2]
    money_revenue = row[6]*row[7]

    if customer in customer_spending:
        customer_spending[customer] += money_revenue
    else:
        customer_spending[customer] = money_revenue

for cus, mon in customer_spending.items():
    print(f"{cus} : ₹ {mon}")
```

```python
# 4th Query : Most payment method used

payment_method = {}

for row in pure_data:
    method = row[9]

    if method in payment_method:
        payment_method[method] += +1
    else: 
        payment_method[method] = +1 

high_used_method = None
times_method = 0

for method, count in payment_method.items():
    if count > times_method:
        times_method = count
        high_used_method = method

for mt, cnt in payment_method.items():
    print(f"{mt} : {cnt}")

print("\nThe popular payment method is")
print(f"{high_used_method} : {times_method}")
```

```python
# 5th Query : Higher revenue generating product

revenue_generating_product = {}

for row in pure_data:
    product = row[4]
    revenue = row[6]*row[7]

    if product in revenue_generating_product:
        revenue_generating_product[product] += revenue
    else:
        revenue_generating_product[product] = revenue

for prd, rev in revenue_generating_product.items():
    print(f"{prd} : ₹ {rev}")

high_revenue_product = None
product_revenue = 0

for prd, rev in revenue_generating_product.items():
    if rev > product_revenue:
        product_revenue = rev
        high_revenue_product = prd

print(f"\nThe higher revenue product is")
print(f"{high_revenue_product} : ₹ {product_revenue}")
```

### Level (3) : Advanced filtering
    Find orders above ₹10,000                       mark(✓)
    Find orders between two dates                   mark(✓)
    Find customers who placed more than 3 orders    mark(✓)
    Calculate discount amount                       mark(✓)
    Find returned orders by category                mark(✓)

```python
# Query 1st : Find order_id above ₹10,000

order_revenue = {}
for row in pure_data:
    revenue = row[6]*row[7]
    oreder_id = row[0]
    if revenue >= 10000:
        order_revenue[oreder_id] = revenue

for ord, rev in order_revenue.items():
    print(f"{ord} : ₹ {rev}")
```

```python
# 2nd Query : Find order b/w two dates

from datetime import datetime
st = input("Enter starting date (YYYY-MM-DD)")
end = input("Enter ending date (YYYY-MM-DD)")

starting_date = datetime.strptime(st, "%Y-%m-%d")
ending_date = datetime.strptime(end, "%Y-%m-%d")

order_by_date = {}

for row in pure_data:
    date = datetime.strptime(row[1], "%Y-%m-%d")
    order_id = row[0]

    if starting_date <= date <= ending_date:
        order_by_date[date] = order_id

for dt, ord in order_by_date.items():
    print(f"{dt} : {ord}")
```

```python
# 3rd Query : Find customer name whose order more than 3.

customer_orders = {}

for row in pure_data:
    customer = row[2]

    if customer in customer_orders:
        customer_orders[customer] += 1
    else:
        customer_orders[customer] = 1

customer_name = {}

for cus, ord in customer_orders.items():
    if ord > 3:
        customer_name[cus] = ord

for cust, ordr in customer_name.items():
    print(f"{cust} : {ordr}")
```

```python
# 4th Query : Calculate Discount amount

discount_amount = {}

for row in pure_data:
    unit_price = row[7]
    discount = float(row[8])
    order_id = row[0]

    if discount == 0:
        discount_amount[order_id] = 0
    else:
        discount_amount[order_id] = unit_price * (discount/100)

for ord, am in discount_amount.items():
    print(f"{ord} :  {am:.2f}")
```

```python
# 5th Query : Returned orders by category wise

order_status = {}
for row in pure_data:
    status = row[10]
    category = row[5]
    if status == "Returned":
        
        if category in order_status:
            order_status[category] += 1
        else:
            order_status[category] = 1
        

for cat, st in order_status.items():
    print(f"{cat} : {st}")

```

```python
total_order = len(pure_data)
total_customer = len(set(row[2] for row in pure_data))
total_product = len(set(row[4] for row in pure_data))

report = open("03_Report.txt","w")

total_quantity = 0
total_revenue = 0

returned_order = 0
pending_order = 0
delivered_order = 0
cancelled_order = 0

upi = 0
cash = 0
card = 0
net_banking = 0

for row in pure_data:
    quantity = int(row[6])
    unit_price = float(row[7])
    discount = float(row[8])
    total_quantity += quantity

    total_revenue += quantity*unit_price * (1 -(discount/100))

    status = row[10]
    payment_mode = row[9]

    if status == "Returned":
        returned_order +=1
    elif status == "Delivered":
        delivered_order +=1
    elif status == "Pending":
        pending_order +=1
    elif status == "Cancelled":
        cancelled_order += 1


    if payment_mode == "UPI":
        upi += 1
    elif payment_mode == "Card":
        card += 1
    elif payment_mode == "Cash":
        cash += 1
    elif payment_mode == "Net Banking": 
        net_banking += 1



report.write("===== RETAIL ORDERS SUMMARY REPORT =====\n\n")

report.write(f"Total Orders: {total_order}\n")
report.write(f"Total Customers: {total_customer}\n")
report.write(f"Total Products: {total_product}\n")
report.write(f"Total Quantity Sold: {total_quantity}\n")
report.write(f"Total Revenue: {total_revenue:.2f}\n\n")

report.write("===== ORDER STATUS =====\n\n")
report.write(f"Returned Orders: {returned_order}\n")
report.write(f"Delivered Orders: {delivered_order}\n")
report.write(f"Pending Orders: {pending_order}\n")
report.write(f"Cancelled Orders: {cancelled_order}\n\n")

report.write("===== PAYMENT_METHOD STATUS =====\n\n")
report.write(f"by UPI: {upi}\n")
report.write(f"by Card: {card}\n")
report.write(f"by Cash: {cash}\n")
report.write(f"by Net Banking: {net_banking}\n")

report.close()

print("Summary report created successfully!")
```

```python
data_revenue = {}


for row in pure_data:
    order_id = row[0]
    quantity = int(row[6])
    unit_price = float(row[7])
    discount = float(row[8])
    
    revenue = quantity * unit_price * (1-(discount/100))
    data_revenue[order_id] = revenue

sorted_revenue = dict(sorted(data_revenue.items(),
    key=lambda x : x[1],
    reverse = True))

sorted_10 = dict(list(sorted_revenue.items())[ :10])

order_details = {}

for row in pure_data:
    order_id = row[0]
    order_details[order_id] = row

import csv

top_file = open("03_Top_10_order.csv", "w", newline="")

writer = csv.writer(top_file)

writer.writerow(["OrderID", "Date", "Customer", "City", "Product","Category", "Quantity", "UnitPrice", "DiscountPct","PaymentMode", "Status", "Revenue"])

for ordr, rev in sorted_10.items():
    row = order_details[ordr]
    writer.writerow(row + [rev])

top_file.close()
print("Your (03_top_10_order.csv) seprate file is ready :")

```
